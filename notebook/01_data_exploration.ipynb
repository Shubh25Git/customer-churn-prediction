## EDA Findings and Insights

During Exploratory Data Analysis (EDA), we examined the relationships between customer features and churn to identify patterns that may help predict customer attrition.

| Feature | Key Observation |
|---|---|
| `tenure` | Customers who churn tend to have shorter tenure with the company. |
| `MonthlyCharges` | Customers who churn pay higher monthly charges on average. |
| `TotalCharges` | Customers who churn have lower total charges on average, likely because they leave earlier. |
| `Contract` | Month-to-month customers have a much higher churn rate than customers with one-year or two-year contracts. |
| `InternetService` | Fiber optic customers have a higher churn rate than DSL customers. |
| `PaymentMethod` | Customers using electronic checks have the highest churn rate among the payment methods analyzed. |
| `OnlineSecurity` | Customers without online security have a higher churn rate than those who have it. |
| `TechSupport` | Customers without technical support have a higher churn rate than those who have it. |
| `PaperlessBilling` | Customers using paperless billing have a higher churn rate than those who do not. |
| `Dependents` | Customers without dependents have a higher churn rate than customers with dependents. |

### Key Takeaways

- **Contract type** shows a strong relationship with customer churn, particularly for month-to-month contracts.
- **Customer tenure** is an important feature to investigate because customers who leave tend to have shorter relationships with the company.
- **Monthly charges and total charges** show different patterns: customers who churn pay more per month on average but accumulate lower total charges.
- **Service-related features**, such as online security and technical support, are associated with differences in churn rates.
- **Payment method** may provide useful information for identifying customers at risk of leaving.

These observations are based on exploratory analysis and do not establish causation. We will evaluate the usefulness of these features during model training and testing.
